Actividad 3 — Estadística descriptiva 

```python
#1. estadisticos, España 1983-2021, media-etc
```

```python
df_1 = dataframe[(dataframe["Entity"] == "Spain") & (dataframe["Year"] >= 1983) &
                  (dataframe["Year"] <= 2021)][["Entity","Year","Annual change in population"]]
df_1.mean()
#df.mean() 
#df.median() 
#df.var() 
#df.std() 
#df.min() 
#df.max() 
```

```python
#2. dataframe original, pais con el minimos y el del maximo, poblacion annual
df_2 = dataframe[(dataframe["Entity"] != "World") & (dataframe["Entity"] != "Less developed regions") &
                 (dataframe["Entity"] != "Less developed regions, excluding China") &
                 (dataframe["Entity"] != "Less developed regions, excluding least developed countries") &
                 (dataframe["Entity"] != "Asia (UN)") & (dataframe["Entity"] != "Lower-middle-income countries") &
                 (dataframe["Entity"] != "Upper-middle-income countries") & (dataframe["Entity"] != "Africa (UN)")
                 & (dataframe["Entity"] != "Least developed countries")
                 ]

df_num2 = df_2[(df_2["Annual change in population"] == df_2["Annual change in population"].min()) |
                     (df_2["Annual change in population"] == df_2["Annual change in population"].max())
                     ][["Entity","Annual change in population"]]
df_num2
```
